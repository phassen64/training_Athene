```python
#Install Required Dependencies
#pip install azure-storage-blob
from azure.storage.blob import BlobServiceClient
import os

# --- CONFIGURATION ---
# PASTE YOUR CONNECTION STRING INSIDE THE QUOTES BELOW
connect_str = "DefaultEndpointsProtocol=https;AccountName=ailabstore998877;AccountKey=PBcOj6w9ez1671hEQlWAV/58+PxIHZNd8imZaPQPjyVffrtwtZNrkl9kvR2GiLVrMRjaDCTcWfCJ+AStAGF6yw==;EndpointSuffix=core.windows.net"

container_name = "lab-container"
local_file_name = "lab-data.txt"
blob_name = "uploaded-lab-data.txt"
download_file_name = "downloaded-lab-data.txt"

# The message to write inside the text file
file_message = "Hello! This is a test file for Azure Blob Storage Lab Practice 6.2."

def main():
    try:
        # 0. Create the local text file with the message (Automation Step)
        print(f"📝 Creating local file '{local_file_name}' with message...")
        with open(local_file_name, "w") as f:
            f.write(file_message)
        print("✅ Local file created successfully.")

        # 1. Create the BlobServiceClient
        print("🔗 Connecting to Azure Storage...")
        blob_service_client = BlobServiceClient.from_connection_string(connect_str)

        # 2. Upload Data to Container
        print(f"⬆️ Uploading {local_file_name} to Azure...")
        blob_client = blob_service_client.get_blob_client(container=container_name, blob=blob_name)
        
        with open(file=local_file_name, mode="rb") as data:
            blob_client.upload_blob(data, overwrite=True)
        print("✅ lab-data.txt Uploaded Successfully")

        # 3. Verify Data (List Blobs in Container)
        print("🔍 Verifying data inside container...")
        container_client = blob_service_client.get_container_client(container=container_name)
        blob_list = container_client.list_blobs()
        print("   Files currently in container:")
        for blob in blob_list:
            print(f"   - {blob.name}")

        # 4. Retrieve (Download) File
        print(f"⬇️ Downloading {blob_name} as {download_file_name}...")
        download_path = os.path.join(os.getcwd(), download_file_name)
        with open(file=download_path, mode="wb") as download_file:
            download_file.write(blob_client.download_blob().readall())
        print("✅ Download successful! Check your folder.")
        
        # Verify content of downloaded file
        with open(download_path, "r") as f:
            content = f.read()
            print(f"   Content verified: '{content}'")

    except Exception as ex:
        print(f"❌ Error occurred: {ex}")

if __name__ == "__main__":
    main()
```
